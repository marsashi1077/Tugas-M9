```python
# nomer 1

waktu = 10

while waktu > 0:
  print(f"waktu ke-{waktu}")
  waktu -= 1
print("waktu selesai!")
```

```python
# nomer 2

bilangan = 2
while bilangan <= 10:
    print(bilangan)
    bilangan += 1 # kode yang kurang
```

```python
# nomer 3

kardus = 0
totalberat = 0

while True:
  kardus += 1
  totalberat += 5
  print(f"kardus ke-{kardus} : {totalberat} kg")
  if totalberat >= 25:
    print ("berat mencapai 25 kg, menghentikan loop!")
    break
```
